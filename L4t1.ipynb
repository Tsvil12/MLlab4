```python
print("Пример 3. График с настройками по умолчанию в Matplotlib:")
def sinplot(flip=1):
    x = np.linspace(0, 14, 100)
    for i in range(1, 5):
        plt.plot(x, np.sin(x + i*.5)*(7 - i)*flip)
sinplot()
plt.show()
```

```python
print("Пример 4. Тот же график, но с настройками Seaborn (sb.set()):")
def sinplot(flip=1):
    x = np.linspace(0, 14, 100)
    for i in range(1, 5):
        plt.plot(x, np.sin(x + i*.5)*(7 - i)*flip)
sb.set()
sinplot()
plt.show()

# Разница в том, что Seaborn автоматически применяет свой стиль (фон, сетку, цветовую палитру),.
```

```python
print("Пример 5. Стиль 'whitegrid':")
def sinplot(flip=1):
    x = np.linspace(0, 14, 100)
    for i in range(1, 5):
        plt.plot(x, np.sin(x + i*.5)*(7 - i)*flip)
sb.set_style("whitegrid")
sinplot()
plt.show()

# Отличие от примера 4 в том, что фон стал белым, а сетка - темной и более четкой.
```

```python
print("Пример 6. Стиль 'white' с удалением верхней и правой осей (despine):")
def sinplot(flip=1):
    x = np.linspace(0, 14, 100)
    for i in range(1, 5):
        plt.plot(x, np.sin(x + i*.5)*(7 - i)*flip)
sb.set_style("white")
sinplot()
sb.despine()
plt.show()
```

```python
print("Пример 8. 'axes.axisbelow': False (сетка поверх графиков):")
def sinplot(flip=1):
    x = np.linspace(0, 14, 100)
    for i in range(1, 5):
        plt.plot(x, np.sin(x + i * .5) * (7 - i) * flip)
sb.set_style("darkgrid", {'axes.axisbelow': False})
sinplot()
sb.despine()
plt.show()

print("Пример 9. 'axes.axisbelow': True (сетка под графиками):")
def sinplot(flip=1):
    x = np.linspace(0, 14, 100)
    for i in range(1, 5):
        plt.plot(x, np.sin(x + i * .5) * (7 - i) * flip)
sb.set_style("darkgrid", {'axes.axisbelow': True})
sinplot()
sb.despine()
plt.show()

# В примере 8 сетка рисуется поверх линий графика, а в примере 9 — под линиями
```

```python
print("Пример 13. Настройка палитры по умолчанию ('husl'):")
def sinplot(flip=1):
    x = np.linspace(0, 14, 100)
    for i in range(1, 5):
        plt.plot(x, np.sin(x + i * .5) * (7 - i) * flip)
sb.set_style("white")
sb.set_palette("husl")
sinplot()
plt.show()
```

```python
print("Пример 14. Диаграмма рассеяния для 'tips':")
sb.set()
tips = sb.load_dataset('tips')
ax = sb.scatterplot(x='total_bill', y='tip', data=tips)
plt.show()

# Зависимость есть. Чем больше счет (total_bill), тем больше чаевые (tip). 
# Характер зависимости — прямая положительная корреляция.
```

```python
print("Пример 15. Гистограмма без KDE:")
df = sb.load_dataset('iris')
sb.distplot(df['petal_length'], kde=False)
plt.show()

print("Пример 16. Только KDE (без гистограммы):")
sb.distplot(df['petal_length'], hist=False)
plt.show()
```

```python
print("Пример 17а. Гистограмма + KDE по умолчанию:")
df = sb.load_dataset('iris')
sb.distplot(df['petal_length'])
plt.show()

print("Пример 17б. С непрозрачностью alpha=1:")
sb.distplot(df['petal_length'], hist_kws={"alpha": 1})
plt.show()
# В варианте б гистограмма стала полностью непрозрачной (альфа=1) и перекрыла линии KDE.
```

```python
print("Пример 18. Двумерное распределение (jointplot):")
df = sb.load_dataset('iris')
sb.jointplot(x='petal_length', y='petal_width', data=df)
plt.show()

# Рассматривается связь между длиной лепестка (petal_length) и его шириной (petal_width).
# Связь существует и она прямая (положительная) — чем длиннее лепесток, тем он шире.
```

```python
print("Пример 19. Парные отношения (pairplot):")
df = sb.load_dataset('iris')
sb.set_style("ticks")
sb.pairplot(df, hue='species', diag_kind="kde", kind="scatter", palette="husl")
plt.show()
```

```python
print("Пример 20. Другой pairplot для тех же данных:")
sb.pairplot(df[df['species'] == 'setosa'], diag_kind="kde", kind="scatter")
plt.show()
```

```python
print("Пример 21. Stripplot без джиттера (jitter=False):")
df = sb.load_dataset('iris')
sb.stripplot(x="species", y="petal_length", data=df, jitter=False)
plt.show()

print("Пример 22. Stripplot с джиттером (jitter=True) — точки разбросаны:")
sb.stripplot(x="species", y="petal_length", data=df, jitter=True)
plt.show()
```

```python
print("Пример 24. Barplot (столбцы со средними значениями):")
df = sb.load_dataset('iris')
sb.barplot(x="species", y="petal_length", data=df)
plt.show()

print("Пример 25. Boxplot (ящик с усами, показывает медиану и выбросы):")
df = sb.load_dataset('iris')
sb.boxplot(x="species", y="petal_length", data=df)
plt.show()

print("Пример 25б. Горизонтальный Boxplot для всех признаков:")
sb.boxplot(data=df, orient="h")
plt.show()
```

```python
print("Пример 26. Violinplot (скрипичный график):")
df = sb.load_dataset('tips')
sb.violinplot(x="day", y="total_bill", data=df)
plt.show()

print("Пример 27. Violinplot с разбивкой по полу (hue='sex'):")
sb.violinplot(x="day", y="total_bill", hue='sex', data=df)
plt.show()
```

```python
print("Пример 28. Barplot для Titanic (выживаемость по полу и классу):")
df = sb.load_dataset('titanic')
sb.barplot(x="sex", y="survived", hue="class", data=df)
plt.show()

print("Пример 29. Countplot (подсчет количества пассажиров по классам):")
sb.countplot(x="class", data=df, palette="Blues")
plt.show()

print("Пример 30. Pointplot (точки со средними значениями):")
sb.pointplot(x="sex", y="survived", hue="class", data=df)
plt.show()
```

```python
print("Пример 31. Регрессия для чаевых (regplot + lmplot):")
df = sb.load_dataset('tips')
sb.regplot(x="total_bill", y="tip", data=df)
sb.lmplot(x="total_bill", y="tip", data=df)
plt.show()

print("Пример 32. Регрессия от размера компании (size) к чаевым (tip):")
sb.lmplot(x="size", y="tip", data=df)
plt.show()

print("Пример 33. Регрессия для датасета Anscombe (набор I):")
df_ans = sb.load_dataset('anscombe')
sb.lmplot(x="x", y="y", data=df_ans.query("dataset == 'I'"))
plt.show()

print("Пример 34. Регрессия для датасета Anscombe (набор II):")
sb.lmplot(x="x", y="y", data=df_ans.query("dataset == 'II'"))
plt.show()
```
